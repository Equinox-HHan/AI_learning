pandas包含各种数据结构和数据操作工具
pandas可以使数据清洗和分析工作更快
pandas专门为异构数据和表格而生

Series是类似于一维数组的对象
它由一组数据以及与之相关的标签组成


```python
import pandas as pd
arr_series=pd.Series([1.0,2.1,1.7,0.8])
print(arr_series)

#可以通过array和index属性获得数组值和索引值
print(arr_series.array)
print(arr_series.index)

#可以指定index
arr_series_index=pd.Series([1,2,3,4],index=["a","b","c","d"])
print(arr_series_index.index)
```

可以通过索引的标签选取Series中的单值或一组值
使用numpy函数或者类似numpy的运算都会保留索引值

```python
import numpy as np

#通过切片进行索引
#不论如何左边永远是index索引
print(arr_series[:],"\n\n")
print(arr_series[1:3],"\n\n")

#进行布尔索引和numpy计算
print(arr_series[arr_series>2],"\n\n")
print(np.exp(arr_series[0:2]))
```

可以将Series看成长度固定的有序字典
本质都是key—value之间的映射关系

```python
key_=["A","B","C","D"]
value_=[1,2,3,4]
#使用zip()函数直接捏合键值
dic=dict(zip(key_,value_))
print(dic)

#将字典和Series互相转化
arr_dict=pd.Series(dic)
print(arr_dict)
dic_arr=arr_series.to_dict()
print(dic_arr)

#如果只传入一个字典则Series index会按照字典键的顺序
#可以将自定义顺序传入构造函数
alp=["C","B","A","E"]
arr_dict_alp=pd.Series(dic,index=alp)
print(arr_dict_alp)
```

可以使用isna()和notna()函数判断是否是NaN
两个函数同样可以作为实例方法
Series的核心功能是计算自动对齐索引标签

```python
ser_A=pd.Series([0.6,0.8,1.0,pd.NA],index=[0,1,2,3])
ser_B=pd.Series([1.7,2.0,2.1,2.4],index=[1,2,3,4])
print(ser_A.isna())
print(pd.notna(ser_B))
print(ser_A+ser_B)
```

Series对象本身和索引index都有name属性
index.name可以原地修改

```python
arr_name=pd.Series([x for x in range(10) if x%2==0])
print(arr_name)
arr_name.name="Changed_Name"
arr_name.index.name="INDEX"
print(arr_name)


arr_name.index=[1,2,3,4,5]
arr_name.name="Changed_Name_2"
print(arr_name)
#手动修改index后如果想要改写name需要重新编辑
arr_name.index.name="INDEX_2"
print(arr_name)
```

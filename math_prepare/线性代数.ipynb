张量是多维的数据数组
一维就是标量
二维就是向量

```python
import torch
x=torch.tensor([2.0])
y=torch.tensor([3.0])
x+y,x*y,x/y,x**y
```

向量可以视为标量组成的列表
可以通过张量的索引访问元素
可以访问张量的长度
只有一个轴的张量形状只有一个元素

```python
import torch
x=torch.arange(4)
print(x)
print(x[3])
print(len(x))
x.shape
```

通过指定m和n创建一个矩阵

```python
import torch
A=torch.arange(20).reshape(5,4)
print(A)
print("\n")
print(A.T)
```

对称矩阵：转置等于自身

```python
import torch
B=torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
print(B==B.T)
```

给定具有相同形状的两个张量
进行二元运算的结果一定是形状相同的张量

```python
import torch
A=torch.arange(20,dtype=torch.float32).reshape(5,4)
B=A.clone()
#clone是深复制，重新分配内存
print(A+B)
print(A*B)
```

计算所有元素的和最后是一个标量
也可以指定某一个维度求和

```python
from sympy import true
import torch
from torch.onnx.symbolic_opset9 import dim
x=torch.tensor(20,dtype=torch.float32)
print(x.sum())
print(x.shape)
y=torch.arange(30).reshape(5,6)
print(y)
print(y.sum(dim=0))
print(y.sum(dim=0,keepdim=True))
```

也可以（在某一个维度上）求均值

```python
import torch
x=torch.arange(20,dtype=torch.float32).reshape(5,4)
print(x.mean())
print(x)
print(x.mean(dim=0,keepdim=True ))
```

点积：按元素相乘最后进行求和

```python
import torch
x=torch.arange(4,dtype=torch.float32)
y=torch.ones(4,dtype=torch.float32)
print(torch.dot(x,y))
```

向量和矩阵相乘以及矩阵和矩阵相乘

```python
import torch
A=torch.tensor([x+2 for x in range(10) if x%2==0])
#只要A是一个1D张量（向量）就不需要转置
#向量在torch中是无方向的
B=torch.arange(30).reshape(6,5)
print(torch.mv(B,A))
C=torch.arange(30).reshape(5,6)
print(torch.mm(B,C))
```

计算L2范数和L1范数

```python
import torch
A=torch.tensor([-3.0,4.0])
#这里必须写入浮点数或者整数dtype=tensor.float
a=torch.norm(A)
b=torch.abs(A).sum()
#曼哈顿距离取绝对值相加
print(a,b)
```

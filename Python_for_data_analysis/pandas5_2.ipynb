讲解Series和DataFrame的基本功能

重建索引是创建一个数据根据新索引排列的新对象
如果某个索引值不存在则会导入缺失值Nan

```python
import pandas as pd

arr_rein=pd.Series([2.0,1.7,2.1,1.0],index=["a","b","c","d"])
arr_rein.reindex(["c","b","d","a","e"])
print(arr_rein)

#需要强调reindex方法是返回新数组而不是原地修改
#因此不完成赋值无法真正reindex
arr_rein_2=arr_rein.reindex(["c","b","d","a","e"])
print(arr_rein_2)
```

对于有序序列做填值处理使用前向填充

```python
import numpy as np

arr_fill=pd.Series(["blue","purple","green"],index=[0,2,4])
print(arr_fill)

#使用method属性和ffill值前向填充
arr_fill_2=arr_fill.reindex(np.arange(6),method="ffill")
print(arr_fill_2)
```

对于DataFrame可以重建行和列或其中单个

```python
arr_df=pd.DataFrame(np.arange(9).reshape((3,3)),index=["a","b","c"],columns=["alph","beta","theta"])
print(arr_df)

#指定重建行或者列
#不存在的行列会返回空缺值
arr_df_2=arr_df.reindex(index=["A","B","C"])
print(arr_df_2)

#重建列索引
col=["gamma","omiga","dirta"]
arr_df_3=arr_df.reindex(columns=col)
#等价于：arr_df_3=arr_df.reindex(col,axis="columns")
print(arr_df_3)

#同时可以使用loc重建
#但是重建的索引必须是已经存在的
#否则不会Nan而是直接KeyError报错
arr_df_4=arr_df.loc[["a","c","b"],["beta","alph","theta"]]
print(arr_df_4)
```

对Series删除指定轴上的项可以重建轴或者使用drop方法
对DataFrame可以使用drop方法（指定关键字或者轴）
需要注意drop方法返回的是一个新的对象

```python
arr1=pd.Series(np.arange(5),index=["A","B","C","D","E"])
print(arr1)
arr1_dp=arr1.drop(["D","E"])
print(arr1_dp)

arr_df_dp1=arr_df.drop(index=["c"])
arr_df_dp2=arr_df.drop(columns=["beta"])
print(arr_df_dp1,arr_df_dp2,sep="\n\n")

arr_df_dp1_1=arr_df.drop("c",axis=0)
arr_df_dp2_2=arr_df.drop("alph",axis=1)
print(arr_df_dp1_1,arr_df_dp2_2,sep="\n\n")
```

Series索引的方式类似numpy数组的索引
要求按照index索引或者使用切片索引
但是推荐使用loc和iloc方法

```python
arr_ser=pd.Series([1,2,3,4],index=["a","b","c","d"])
print(arr_ser["a"],arr_ser[1:],sep="\n\n")

#loc运算符只使用标签iloc运算符只使用整数下标
#无论索引是否包含整数都能使用iloc
print(arr_ser.loc[["a","c"]])
#loc切片是包含尾端的
#使用该方法还能进行赋值
print(arr_ser.loc["a":"c"])

arr_ser_index=arr_ser.reindex([100,200,300])
print(arr_ser_index.iloc[np.array([2,0])])
```

用单个值或者切片对DataFrame进行索引还可以使用布尔索引

```python
data=pd.DataFrame(np.arange(16).reshape((4,4)),index=["A","B","C","D"],columns=["alpa","beta","omiga","sigma"])
print(data["alpa"],data[["beta","sigma"]],sep="\n\n")
print(data[1:3],data[data["alpa"]>10],sep="\n\n")

print(data<8)
#data[data<8]=0
#这是一个赋值语句不能直接被print
```

对DataFrame使用loc进行单值索取会取出Series使用切片会取出DataFrame
使用iloc可以做出同样类似的索引,df的所有iloc方法都是先行后列

```python
print(data.loc["A"],data.loc[["C","D"]],sep="\n\n")
#选取行和列需要使用逗号进行分隔
print(data.loc["A",["omiga","sigma"]])

print(data.iloc[2],data.iloc[[1,3]],data.iloc[1,[2,3]],sep="\n\n")
```

对于Series不重叠标签数据对齐会导入缺失值并且缺失值会在以后的算术过程中传播
对于DataFrame对齐操作会同时发生在行和列上取并集
带有填充值的特殊方法fill_value=A表示将所有nan替换为A后做数学运算

```python
s1=pd.Series([7.3,-2.5,3.4,1.5],index=["a","c","d","e"])
s2=pd.Series([-2.1,3.6,-1.5,4,3.1],index=["a","c","e","f","g"])
print(s1+s2)

df1=pd.DataFrame(np.arange(9.).reshape((3,3)),index=["Ohio","Texas","Colorado"],columns=list("bcd"))
df2=pd.DataFrame(np.arange(12.).reshape((4,3)),index=["Utah","Ohio","Texas","Oregon"],columns=list("bde"))
print(df1,df2,df1+df2,sep="\n\n")

df3=pd.DataFrame(np.arange(12.).reshape((3,4)),columns=list("abcd"))
df4=pd.DataFrame(np.arange(20.).reshape(4,5),columns=list("abcde"))
print(df3,df4,sep="\n\n")
df=df3.add(df4,fill_value=0)
print(df)
```

DataFrame和Series之间的运算
提示：numpy运算具有广播机制

```python
frame=pd.DataFrame(np.arange(12.).reshape((4,3)),columns=list("bde"),index=["Utah","Ohio","Texas","Oregon"])
series=frame.iloc[0]
print(frame,series,sep="\n\n")

#默认情况下Series的索引会匹配到DataFrame的列然后进行广播
print(frame-series)

#如果索引值在Dataframe的列或者series的索引中找不到
#则参与运算的对象会重建索引并且形成并集
series2=pd.Series(np.arange(3),index=["b","e","f"])
print(series2,frame+series2,sep="\n\n")

#如果希望在列上进行广播则需要指定轴
series3=frame["d"]
print(series3)
print(frame.sub(series3,axis=0))
```

函数应用和映射：numpy的通用函数也可以应用于pandas对象
将操作应用到行列组成的一维数组上可以使用apply方法

```python
frame2=pd.DataFrame(np.random.standard_normal((4,3)),columns=list("bde"),index=["Utah","Ohio","Texas","Oregon"])
print(frame2,abs(frame2),sep="\n\n")

def func(x):
    return x.max()-x.min()
frame3=frame2.apply(func,axis="index")
frame4=frame2.apply(func,axis="columns")
print(frame3,frame4,sep="\n\n")

#apply不止能够单个标量值还能返回Series
def func2(x):
    return pd.Series([x.min(),x.max()],index=["min","max"])
frame5=frame.apply(func2)
```

对于Series的排序操作：按键排序sort_index()按值排序sort_value()
在排序时任何缺失值都会放到末尾也可以使用na_position参数

```python
series1=pd.Series(np.arange(4),index=["b","a","c","d"])
series1_1=series1.sort_index()
print(series1_1)

series2_2=pd.Series([4,np.nan,7,np.nan,-3,2])
series2_3=series2_2.sort_values()
print(series2_3)
```

对于Dataframe的排序操作：按轴排序指定轴还可以使用ascending参数升降序
将列名传递给排序函数即可按值排序

```python
frame1=pd.DataFrame(np.arange(8).reshape((2,4)),index=["three","one"],columns=["d","a","b","c"])
frame1_1=frame1.sort_index(axis=0)
frame1_2=frame1.sort_index(axis=1,ascending=False)
print(frame1_1,frame1_2,sep="\n\n")

frame1_3=frame1.sort_values(["b"])
print(frame1_3)
```

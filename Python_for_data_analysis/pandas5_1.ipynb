Series是类似于一维数组的对象
它由一组数据以及与之相关的标签组成


```python
import pandas as pd
arr_series=pd.Series([1.0,2.1,1.7,0.8])
print(arr_series,"\n\n")

#可以通过array和index属性获得数组值和索引值
print(arr_series.array,"\n\n")
print(arr_series.index,"\n\n")

#可以指定index
arr_series_index=pd.Series([1,2,3,4],index=["a","b","c","d"])
print(arr_series_index.index,"\n\n")
print(arr_series_index)
```

可以通过索引的标签选取Series中的单值或一组值
使用numpy函数或者类似numpy的运算都会保留索引值

```python
import numpy as np

#通过切片进行索引
#不论如何左边永远是index索引
print(arr_series[:],"\n\n")
print(arr_series[1:3],"\n\n")

#进行布尔索引和numpy计算
print(arr_series[arr_series>2],"\n\n")
print(np.exp(arr_series[0:2]))
```

可以将Series看成长度固定的有序字典
本质都是key—value之间的映射关系

```python
key_=["A","B","C","D"]
value_=[1,2,3,4]
#使用zip()函数直接捏合键值
dic=dict(zip(key_,value_))
print(dic)

#将字典和Series互相转化
arr_dict=pd.Series(dic)
print(arr_dict)
dic_arr=arr_series.to_dict()
print(dic_arr)

#如果只传入一个字典则Series index会按照字典键的顺序
#可以将自定义顺序传入构造函数
alp=["C","B","A","E"]
arr_dict_alp=pd.Series(dic,index=alp)
print(arr_dict_alp)
```

可以使用isna()和notna()函数判断是否是NaN
两个函数同样可以作为实例方法
Series的核心功能是计算自动对齐索引标签

```python
ser_A=pd.Series([0.6,0.8,1.0,pd.NA],index=[0,1,2,3])
ser_B=pd.Series([1.7,2.0,2.1,2.4],index=[1,2,3,4])
print(ser_A.isna())
print(pd.notna(ser_B))
print(ser_A+ser_B)
```

Series对象本身和索引index都有name属性
index.name可以原地修改

```python
arr_name=pd.Series([x for x in range(10) if x%2==0])
print(arr_name)
arr_name.name="Changed_Name"
arr_name.index.name="INDEX"
print(arr_name)


arr_name.index=[1,2,3,4,5]
arr_name.name="Changed_Name_2"
print(arr_name)
#手动修改index后如果想要改写name需要重新编辑
arr_name.index.name="INDEX_2"
print(arr_name)
```

DataFrame是矩形的数据表具有有序且命名的列
其中每一列都可以使用不同的数据类型

```python
#创建DataFrame传入等长列表或者numpy数组构成的字典
#生成的DataFrame会自动生成索引
data={"name":["xiaohan","xiaoming","xiaobai","xiaochen"],"age":[20,21,10,17],"score":[100,95,90,99]}
frame=pd.DataFrame(data)
print(frame)
```

使用head()和tail()方法会返回较大DataFrame的前五行和后五航
可以指定列的排序顺序（如果字典没有指定的列会产生缺失值）
通过标记法或者点属性可以使DataFrame返回一个Series

```python
frame_col=pd.DataFrame(data,columns=["score","name","age"])
print(frame_col,"\n\n")
frame_col2=pd.DataFrame(data,columns=["score","name","age","performance"])
print(frame_col2,"\n\n")

#检查各种属性
print(frame_col2.shape)
print(frame_col2.dtypes,"\n\n")

#强烈建议使用括号索引而不是点属性
#点属性对列名有要求且不能和DataFrame方法冲突
print(frame_col["name"])
```

使用iloc和loc方法可以获取行属性
可以直接通过赋值的方法修改列内容

```python
print(frame_col2.loc[0])
frame_col2["performance"]="A"
print(frame_col2)
print(frame_col2.loc[0])
```

将列表或者数组赋值给某列时长度必须相等
如果赋值内容是Series则index会对应锁定
为不存在的列赋值会重新插入一个列
建议使用.drop()方法删除某列

```python
per=pd.Series(["A","B","A","C"],index=[0,2,1,3])
frame_col2["performance"]=per
frame_col2["isPassed"]=True
print(frame_col2)

frame_col2=frame_col2.drop(columns=["age"],errors="ignore")
print(frame_col2)
```

通过索引从DataFrame返回的列是视图任何修改都会直接作用到DataFrame上
将嵌套字典传给DataFrame会将外层key解释为列内层key解释为行索引

```python
population={"Ohio":{2000:1.5,2001:1.7,2002:3.6},"Nevada":{2001:2.4,2002:2.9}}
frame_pop=pd.DataFrame(population)
print(frame_pop)

#使用.T方法可以转置DataFrame
print(frame_pop.T)
```

关于如何在工业和科研中写出一段高效的并且优雅的转置代码
错误的转置会导致内存暴涨向量化计算效率下降和某些统计方法出错

```python
import pandas as pd
import numpy as np

data_beforeT={"sample_id":["Sample_A","Sample_B","Sample_C"],
      "gene_1":[12.5,14.1,10.8],"gene_2":[100,105,98],"score":[1,2,3]}
df=pd.DataFrame(data_beforeT)

#错误做法：直接转置导致类型全部变成object
#正确做法：把标识列设置为index后转置
df_clean=df.set_index("sample_id").T

print("---DataFrame---")
print(df)
print("\n---直接转置---")
print(df.T)

print("\n---转置后的DataFrame---")
print(df_clean)
print("\n---检查数据类型---")
print(df_clean.dtypes)
print("\n---计算样本均值---")
print(df_clean.mean())
```

```python
import pandas as pd

# 1. 模拟财务/用户宽表数据
df_wide = pd.DataFrame({
    'user_id': [1001, 1002],
    'region': ['East', 'West'],
    'Jan_sales': [150.5, 200.0],
    'Feb_sales': [180.0, 210.5],
    'Mar_sales': [210.0, 250.0]
})

print(df_wide)
# ✅ 工业标准做法：用 melt 进行“逆透视/长表转换”
df_long = df_wide.melt(
    id_vars=['user_id', 'region'],                    # 保持不动的列
    value_vars=['Jan_sales', 'Feb_sales', 'Mar_sales'],# 需要解包/转置的列
    var_name='month',                                 # 拆开后的列名
    value_name='sales_amount'                         # 对应的数值列名
)

print("--- 重构后的长表（符合 Tidy Data 标准）---")
print(df_long)
```

DataFrame的index和columes有name属性但是DataFrame本身没有

```python
data_non={"A":{"a":"vscode","b":"codex","c":"zcode"},
          "B":{"d":"claude","e":"chrome","f":"clash"}}
#外层key作为columes
#内层key作为index
#不存在的对应显示NaN
data_f=pd.DataFrame(data_non)
print(data_f)


data_f.index=list(range(6))
data_f["A"]=np.nan
data_f["B"]=pd.Series(["OpenAI","Google","DeepSeek","Moonshot","Anthropic","Microsoft"])
data_f.index.name="INDEX"
data_f.columns.name="App"
print(data_f)
```
